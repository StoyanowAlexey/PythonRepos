## Завдання 1
Введіть список, що складається з *n* цілих елементів. Знайдіть кількість парних елементів. *n* вводиться з клавіатури.

```python
def count_even(a):
    count = 0
    for i in range(len(a)):
        if a[i] % 2 == 0:
            count = count + 1
    print(count)

n = int(input("n = "))
text = input("List all: ")
parts = text.split()

a = []
for i in range(n):
    a.append(int(parts[i]))

count_even(a)
```

```python
count_even([9, 0, 1, 2, 3, 4, 5, 9])
count_even([0, 0, 0, 0])
count_even([])
count_even([1, 3, 5, 7, 9])
```

## Завдання 2
Введіть список, в якому є лише два однакових елемента. Визначте їх індекси.

```python
def same_indexes(a):
    for i in range(len(a)):
        for j in range(i + 1, len(a)):
            if a[i] == a[j]:
                print(str(i) + ", " + str(j))
                return

text = input("List: ")
parts = text.split()

a = []
for p in parts:
    a.append(int(p))

same_indexes(a)
```

```python
same_indexes([45, 99, 66, 32, 67, 78, 99, 10])
same_indexes([0, 0, 2, 4, 6, 7, 9, 1])
same_indexes([9, 0, 1, 2, 3, 4, 5, 9])
```

## Завдання 3
Видаліть зі списку всі елементи перед тим, що задається в якості параметра.

У нас є два граничних випадки: (1) якщо елемент, що вводиться, неможливо
знайти, то список не слід змінювати. (2) якщо список порожній, він повинен
залишатися порожнім.

```python
def remove_before(a, x):
    start = 0
    for i in range(len(a)):
        if a[i] == x:
            start = i
            break

    result = []
    for i in range(start, len(a)):
        result.append(a[i])

    print(result)

text = input("List (with space between numbers): ")
parts = text.split()

a = []
for p in parts:
    a.append(int(p))

x = int(input("Enter element: "))
remove_before(a, x)
```

```python
remove_before([1, 2, 3, 4, 5], 3)
remove_before([1, 1, 2, 2, 3, 3], 2)
remove_before([1, 1, 2, 4, 2, 3, 4], 2)
remove_before([1, 1, 5, 6, 7], 2)
remove_before([], 2)
```
