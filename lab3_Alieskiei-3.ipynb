## Завдання 1.1
Напишіть функцію, яка приймає ім'я, по батькові та прізвище, та повертає рядок, де ім'я та по батькові є ініціалами.

Приклад: Володимир Олександрович Зеленський → В. О. Зеленський

```python
def initials(name, patronymic, surname):
    result = name[0] + ". " + patronymic[0] + ". " + surname
    return result

initials("Володимир", "Олександрович", "Зеленський")
```

```python
assert initials("Володимир", "Олександрович", "Зеленський") == "В. О. Зеленський"
assert initials("Тарас", "Григорович", "Шевченко") == "Т. Г. Шевченко"
assert initials("Іван", "Якович", "Франко") == "І. Я. Франко"
assert initials("Алєксєй", "Олександрович", "Стоянов") == "А. О. Стоянов"
assert initials("James", "Herbert", "Bond") == "J. H. Bond"
print("All tests passed")
```

## Завдання 1.2
Напишіть функцію, яка приймає три параметри: словник (d), старий ключ елемента та новий ключ елемента. Функція повертає словник d зі зміненим ключем.

```python
def change_key(d, old_key, new_key):
    result = {}
    for key in d:
        if key == old_key:
            result[new_key] = d[key]
        else:
            result[key] = d[key]
    return result

dict1 = {}
for i in range(1, 11):
    dict1[str(i)] = i
dict1["total"] = 55

change_key(dict1, "total", "sum")
```

```python
assert change_key({"total": 55}, "total", "sum") == {"sum": 55}
assert change_key({"a": 1, "b": 2}, "a", "c") == {"c": 1, "b": 2}
assert change_key({"x": 1, "total": 3}, "total", "sum") == {"x": 1, "sum": 3}
assert change_key({"a": 1}, "x", "y") == {"a": 1}
assert change_key({}, "a", "b") == {}
print("All tests passed")
```

## Завдання 2.1
Для тексту в «text.txt» підрахуйте, скільки разів кожне слово зустрічається в тексті, і сформуйте словник з елементами <word> : <скільки разів>.

Перед запуском завантажте файл **text.txt** у Colab: панель «Файли» зліва => «Завантажити до сховища сеансу».

```python
with open("text.txt") as file:
    text = file.read()
```

```python
def count_words(text):
    text = text.lower()
    text = text.replace("'s", "")
    text = text.replace("’s", "")
    text = text.replace("\n", " ")

    symbols = [".", ",", "(", ")", "[", "]", "{", "}"]
    for s in symbols:
        text = text.replace(s, "")

    words = text.split()

    d = {}
    for word in words:
        if word in d:
            d[word] = d[word] + 1
        else:
            d[word] = 1
    return d

def print_dict(d):
    for key in d:
        print(key, ":", d[key])

d = count_words(text)
print_dict(d)
```

```python
assert count_words("a b a") == {"a": 2, "b": 1}
assert count_words("Hello, hello.\nWorld") == {"hello": 2, "world": 1}
assert count_words("Python's (best) list") == {"python": 1, "best": 1, "list": 1}
assert count_words("") == {}
print("All tests passed")
```

## Завдання 2.2
Створіть функцію, яка знаходить найуживаніше слово в тексті.

```python
def most_common_word(d):
    best_word = ""
    best_count = 0
    for word in d:
        if d[word] > best_count:
            best_word = word
            best_count = d[word]
    return best_word

word = most_common_word(d)
print(word, ":", d[word])
```

```python
assert most_common_word({"a": 2, "b": 5, "c": 1}) == "b"
assert most_common_word({"and": 20, "the": 17, "python": 19}) == "and"
assert most_common_word({"one": 1}) == "one"
assert most_common_word({}) == ""
print("All tests passed")
```
