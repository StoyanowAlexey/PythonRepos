```python
def is_prime(n):
    """Checks if a number is prime."""
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def task_6():
    user_input = input("Enter 2 numbers a and b: ")

    try:
        parts = user_input.split()
        if len(parts) != 2:
            return "Error. You must enter exactly two numbers separated by a space."

        a = int(parts[0])
        b = int(parts[1])

    except ValueError:
        return "Error. A non-numeric value is present in the list."

    start = min(a, b)
    end = max(a, b)

    primes = [num for num in range(start, end + 1) if is_prime(num)]
    return primes

print("In [2]: task_6() # Please enter: 3 15")
print(f"Out[2]: {task_6()}\n")

print("In [3]: task_6() # Please enter: 6 24")
print(f"Out[3]: {task_6()}\n")

print("In [4]: task_6() # Please enter: 3 str")
print(f"Out[4]: {task_6()}")
```
